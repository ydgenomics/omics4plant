## Preprocessing

```python
data = snap.pp.imort_fragment(
    fragment_file,
    chrom_size=snap.genome.hg38,
    file="pbmc.h5ad", # optional
    sorted_by_barcode=False,
)
```

`pl.frag_size_distr()` to calculate and plot the size distribution of fragments in this dataset
- Nucleosome-Free Region (NFR)
- Mono-Nucleosome Peaks
- Di-Nucleosome Peaks
- Large Fragments

```python
snap.pl.frag_size_distr(data, interactive=False)
```

```python
fig = snap.pl.frag_size_distr(data, show=False)
fig.update_yaxes(type="log")
fig.show()
```

TSS enrichment (TSSe): This is often indicative of open and accessible promoter regions, which are typically devoid of nucleosomes and are involved in active gene transcription.

```python
snap.metrics.tsse(data, snap.genome.hg38)
```

```python
snap.pl.tsse(data, interactive=False)
```
