Firstly, download our [example data](http://upload.dcs.cloud:8090/share/bb6fab82-2c16-46b2-a95e-6931338f31bf) and import Stereopy.

```python
import stereo as st
import warnings
warnings.filterwarnings('ignore')
```

```python
sample='wt2'
```

Get the version information of Stereopy.

```python
# data_path = '/data/input/Files/yangdong/M.truncatula/SAW/WT202604020036551/result/Y00710F6/outs/feature_expression/Y00710F6.tissue.gef'
data_path='/data/input/Files/yangdong/M.truncatula/SAW/WT202604020036551/result/Y00710F6/outs/feature_expression/Y00710F6.tissue.gef'
st.io.read_gef_info(data_path)

data = st.io.read_gef(file_path=data_path, bin_size=80)

# simply type the varibale to get related information
data

data.tl.cal_qc()

ins = data.plt.interact_spatial_scatter()

ins.show()
```

Load data to generate a StereoExpData object.

```python
data1 = ins.selected_exp_data
print(data1)

data1.cells['sample'] = sample
data1.tl.cal_qc()

data1.plt.violin()

data1.plt.spatial_scatter()

marker_list = ["Chr2012264","Chr4022018","Chr4023437","Chr4026732","Chr4027514","Chr5031033","Chr5031730"]

# 筛选出存在于 real_gene_name 列中的基因
valid_markers = [
    gene for gene in marker_list 
    if gene in data1.genes['real_gene_name'].values
]

# 打印一下，确认过滤掉了哪些
print(f"原始数量: {len(marker_list)}")
print(f"过滤后有效数量: {len(valid_markers)}")
print(f"有效基因: {valid_markers}")


data1.plt.spatial_scatter_by_gene(gene_name=valid_markers)

print(data1)

adata1 = st.io.stereo_to_anndata(data1,flavor='scanpy',output=sample + '_scanpy_out.h5ad')
adata1 = st.io.stereo_to_anndata(data1,flavor='seurat',output=sample + '_seurat_out.h5ad')
```
